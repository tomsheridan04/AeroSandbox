# APC 12x8E Direct + Root Chord+Twist Optimization

This notebook runs the chord+twist optimization from `apc_12x8e_direct_root_chord_twist_optimization.py`. Chord and twist are B-spline design variables. The optimized thickness ratio is computed from the optimized chord so that `chord * t/c` matches the baseline APC absolute thickness at every station.

```python
%matplotlib inline

from pathlib import Path
import sys
import importlib

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / "aerosandbox").exists() and (candidate / "studies").exists():
            return candidate
    raise RuntimeError("Could not find the AeroSandbox repository root.")

REPO_ROOT = find_repo_root()
sys.path.insert(0, str(REPO_ROOT))

from studies.PropellerAnalysisValidation import apc_12x8e_direct_root_chord_twist_optimization as opt
opt = importlib.reload(opt)
```

## Editable Settings

```python
opt.DESIGN_RPM = 5000
opt.DESIGN_VELOCITY_MPH = 10.5
opt.DESIGN_VELOCITY_MPS = opt.DESIGN_VELOCITY_MPH * 0.44704
opt.DESIGN_LABEL = "5000 RPM, 10.5 mph low-J climb"

opt.N_CHORD_CONTROL_POINTS = 6
opt.N_TWIST_CONTROL_POINTS = 6
opt.SPLINE_DEGREE = 3

opt.OPTIMIZATION_RADIAL_RESOLUTION = 10
opt.REPORT_RADIAL_RESOLUTION = 24
opt.CHECK_RADIAL_RESOLUTION = 32

opt.THRUST_MATCH_FACTOR = 1.0
opt.THRUST_TARGET_MARGIN_N = 0.12

opt.MIN_CHORD_M = 1e-4
opt.MAX_CHORD_M = 0.12
opt.MIN_THICKNESS_RATIO = 0.055
opt.MAX_THICKNESS_RATIO = 0.42

opt.CHORD_TWIST_OPTIMIZATION_OUTPUTS = (
    opt.OUTPUTS / "chord_twist_optimization_free_chord_direct_root_low_j"
)
```

## Run Optimization

```python
summary, control_points, geometry_table = opt.run_chord_twist_optimization()
summary
```
